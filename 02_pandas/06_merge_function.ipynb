Question: Find the Orange Cap winner for each season (the player who scored the most runs in that season).

```python
delivery = pd.read_csv("deliveries.csv")
matches=pd.read_csv("matches - matches.csv")
```

```python
new=delivery.merge(matches,left_on='match_id',right_on='id')
```

```python
# Now we will use groupby() on the 'season' and 'batsman' columns.
new.groupby(['season','batsman'])['batsman_runs'].sum()
```

The series we obtain has 2 index (season and batsman) and 1 value is runs .
this type of series is known as Multi Indexed Series 

Now we will convert this Series into a DataFrame using the reset_index() function. (The opposite function is set_index().) Before doing that, we will sort the values in descending order (ascending=False) so that the players with the highest runs appear first. Although sorting changes the order of the seasons, that's not a problem. We can use the drop_duplicates() method to keep only the first player (the highest run-scorer) for each season and remove the remaining players.


```python
new.groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().drop_duplicates(subset=['season'],keep='first').sort_values('season')[['season','batsman']]
```
