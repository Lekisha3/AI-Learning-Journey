```python
import numpy as np 
import pandas as pd 
```

```python
data = pd.read_csv("matches - matches.csv")
```

The sort_values() method :

```python
# In Series 
(data['team2'].value_counts()+data['team1'].value_counts()).sort_values(ascending=False)
```

```python
# In dataframe
data.sort_values('city',ascending=False)
```

```python
# For permanent changes 
data.sort_values('city',ascending=False,inplace=True)
```

```python
data
```

```python
#Applying sort_values on multiple columns 
data.sort_values(['city','date'])
```

```python
#Question : Now what if we want city in ascending order and date in descending order 
data.sort_values(['city','date'],ascending=[True,False])
```

```python
data['winner'].value_counts().sort_values(ascending=False)
```

The drop_duplicates() method :
When we working on machine learning algorithms, so we don't give duplicate data to algorithm , it reduces efficiency. 
So Pandas has drop_duplicate method , to remove duplicate data

```python
remove_duplicate=data.drop_duplicates(subset=['city'])
```

```python
remove_duplicate
```

```python
#drop_duplicate on multiple column 
remove=data.drop_duplicates(['city','season'])
```
