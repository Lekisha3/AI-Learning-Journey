```python
# Find the batsman who hit the most fours in the dataset. 
mask=delivery['batsman_runs']==4
new_delivery=delivery[mask]

new_delivery.shape
```

```python
new_delivery.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head(5)
```

```python
# Find the team against which Virat Kohli has scored the most runs. 
mask=delivery['batsman']=='V Kohli'
vk=delivery[mask]
vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head()
```

```python
# we can also create function
def runs_scored(batsman_name):
    runs=delivery[delivery['batsman']==batsman_name]
    return runs['batsman_runs'].sum()

runs_scored('V Kohli')
```
