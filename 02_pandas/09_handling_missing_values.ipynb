Handling Missing Values: 
Has 2 ways :-
1. Dropping missing data using dropna()
2. Filling missing values using fillna() 

We cannot pass missing values to an ML algorithm because they can confuse the algorithm .

```python
#dropna() method 
import numpy as np
import pandas as pd 
```

```python
data=pd.read_csv('train.csv')
data.head()
```

```python
data.shape
```

```python
data.dropna()    
```

```python
data.dropna(axis=1).shape          #Now it will delete any column that has even a single missing value.
```

```python
data.dropna(subset=['Cabin']).shape   # We can specify multiple columns. The function only checks the columns we mention. If any of those columns has a missing value, it removes that row.
```

```python
data.dropna(subset=['Cabin','Embarked']).shape
```

2. fillna() method :-
Sometimes, a row or column is important, so we don't want to delete it just because it has missing values. Instead, we use the fillna() method to fill the missing values.

```python
data.fillna(0)
```

Suppose we have two columns: one contains integers and the other contains strings. Filling missing values with 0 is not suitable for both columns. Instead of applying fillna() to the entire DataFrame, we can apply it to a specific column.

```python
data['Cabin'].fillna('Not Specified')
```

The fillna() function also provides several options for filling missing values.

```python
data['Age']=data['Age'].bfill()
```
