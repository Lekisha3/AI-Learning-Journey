Scatter Plots :-

1. Bivariate Analysis (Analyzing Two Columns Together)
2. Numerical vs Numerical (Mostly)
3. Use Case: Finding correlation between 2 quantities 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns  #to use for dataset
```

```python
#plt.scatter simple function 
x=np.linspace(-10,10,50)  # linspace function : creates evenly spaced numbers between a start value and an end value 
x
y=10*x+3+np.random.randint(0,300,50)
y

plt.scatter(x,y)
```

```python
# plt.scatter on pandas data
df=pd.read_csv('batter.csv')
df.head()
```

```python
df.head(50)
plt.scatter(df['avg'],df['strike_rate'])
plt.title('Avg and SR analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red',marker='+')
plt.title('Avg and SR analysis of top 50 batsman')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
```

```python
# Size 
tips=sns.load_dataset('tips')
tips
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
```

```python
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20) 

'''The size column in the dataset represents the number of people a person brought along.
When we use this column in the plot, the more people a person brought, the larger the corresponding dot (marker) appears.
We multiply the values by 20 because the original values are very small, making the markers easier to see on the graph.'''
```

```python
# scatter plot using plt.plot
plt.plot(tips['total_bill'],tips['tip'])

# output is 2D plot
```

```python
plt.plot(tips['total_bill'],tips['tip'],'o')

# output is scatter plot
```

However, if we create a scatter plot using plt.plot(), we cannot use the size parameter, and color manipulation is also difficult.

So, why use plt.plot() to create a scatter plot?

1. plt.plot() is generally faster, whereas plt.scatter() is comparatively slower.
2. If you are working with a very large dataset, plt.plot() is often the better choice because of its speed. However, if you are working with a smaller dataset and need greater control over size, color, and other visual properties, use plt.scatter().
