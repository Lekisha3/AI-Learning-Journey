Pie Chart :-

Plot over categories generally shows how much each category contributes to the total (100%).

1. Univariate/Bivariant Analysis
2. Categorical Vs Numerical
3. Use Case:- To find contribution on a standard scale.

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns  #to use for dataset
```

```python
# Simple data 
data=[23,45,100,98,76,20]
plt.pie(data)
plt.show()
```

It is not necessary for all the numbers to add up to 100. Whatever the total sum is, the percentage is calculated based on that total. For example, if one value is 23, the graph shows what percentage 23 contributes to the overall total.

```python
# labels 
data=[23,45,100,20,49]
subjects=['eng','hindi','maths','science','sst']
plt.pie(data,labels=subjects)
plt.show()
```

```python
# Dataset practice
df=pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'])
```

Problem: The percentage (%) is not being displayed.
Solution:For this, we have one more parameter.

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
```

```python
# Colors 
'''We can also specify the colors. Previously, the colors were generated automatically.'''
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','pink','yellow','red','orange'])
```

```python
# Explode Shadow 
'''Enhance the look of the graph.
 We can also extract a particular slice of the pie chart.''' 
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0])
```

We can assign a value to the slice(s) that we want to pull out and set the other slices to 0. This can be done for one or more slices. The amount by which the slice is pulled out depends on the value we specify. For example, instead of 0.1, if we use 0.3, the slice will be pulled out further.

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0])
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0],shadow=True)
```
