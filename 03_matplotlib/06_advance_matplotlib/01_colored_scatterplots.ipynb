```python
# Now if I want to see correlation between 2 column 
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

In this dataset, each row represents one flower, which belongs to one of three categories (varieties).

Now, we want to separate these varieties using different colors in the scatter plot. For this, we will assign a numerical value to each species.

```python
iris=pd.read_csv('iris.csv')
iris.sample(5)
iris['Species']=iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

The colors are generated automatically, but we can also define the colors ourselves.

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
# We can also add 
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()   # Define which color represents which value.
```

```python
# Alpha — controls the opacity of the graph.
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet',alpha=0.5)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

Alpha: A lower value gives the graph a more translucent look. Its value can range from 0 to 1.

```python
# Plot Size (Applicable on all graphs)
plt.figure(figsize=(15,7)) # here 15 is width and 7 is height
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'],cmap='jet')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```
