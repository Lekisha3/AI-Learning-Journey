Subplots:-
When we are working with multiple columns and want to plot their graphs side by side instead of separately, so that we can easily compare the differences and see what is going on, we can use subplots.

A Different way to plot graphs:-

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
batters=pd.read_csv('batter.csv')
batters.head()
```

```python
# 1st 
plt.scatter(batters['avg'],batters['strike_rate'])
```

```python
# 2nd (The main reason for using subplots is to plot multiple graphs together in a single figure)
fig,ax=plt.subplots()
ax.scatter(batters['avg'],batters['strike_rate'])
```

Comparison of both methods

```python
# 1st
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Analysis')
plt.xlabel('Average')
plt.ylabel('Strike Rate')
plt.show()
```

```python
# 2nd 
fig,ax=plt.subplots(figsize=(15,6))
ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title('Analysis')
ax.set_xlabel('Average')
ax.set_ylabel('Strike Rate')
```

So, now I want to plot two graphs side by side: one showing the relationship between average and strike rate, and the other showing the relationship between runs and average. In other words, I want to create two scatter plots side by side.


```python
fig,ax=plt.subplots(nrows=2,ncols=1)
ax[0].scatter(batters['avg'],batters['strike_rate'])
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[1].set_title('Avg Vs Runs')

ax[0].set_ylabel('Strike Rate')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

Here, we can observe that the x-axis points of the upper graph are merging with the title of the lower graph. To avoid this, we can use shared x-axis for both graphs.

```python
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True)
ax[0].scatter(batters['avg'],batters['strike_rate'])
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[1].set_title('Avg Vs Runs')

ax[0].set_ylabel('Strike Rate')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
# We can also change the color.
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True)
ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[1].set_title('Avg Vs Runs')

ax[0].set_ylabel('Strike Rate')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
# We can also increase the size
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))
ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[1].set_title('Avg Vs Runs')

ax[0].set_ylabel('Strike Rate')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

Now, plot 4 graphs 

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,10))
ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```
