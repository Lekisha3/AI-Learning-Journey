1. 3D Scatter plots 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
batters=pd.read_csv('batter.csv')
batters.head()
```

```python
fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'])
ax.set_title('IPL batsman analysis')
ax.set_xlabel('Runs')
ax.set_ylabel('avg')
ax.set_zlabel('Strike Rate')
```

2. 3D Line Plot

```python
x=[0,1,5]
y=[0,10,13]
z=[0,13,20]

fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(x,y,z,s=[100,100,100])   
ax.plot3D(x,y,z,color='red')
```

3. 3D Surface Plots 

```python
# Plot graph of 3d function : f(x,y)=x**2+y**2
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
xx.shape
z=xx**2 + yy**2
z.shape

fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z)
```

```python
# We can also change the color
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
xx.shape
z=xx**2 + yy**2
z.shape

fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# Eg:- z=np.sin(xx)+np.cos(yy)
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)
xx,yy=np.meshgrid(x,y)
xx.shape
z=np.sin(xx)+np.cos(yy)
z.shape

fig=plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

4. Contour Plots 
A special type of plot that represents a 3D graph in 2D.

```python
z=xx**2+yy**2
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z=np.sin(xx)+np.cos(yy)
fig=plt.figure(figsize=(12,8))
ax=plt.subplot()
p=ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
