```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
batters=pd.read_csv('batter.csv')
batters.head()
```

```python
sample_df=batters.head(100).sample(25,random_state=5)
```

```python
# If I want to identify batsmen whose strike rate is more than 130, we can draw a horizontal line at 130 to easily identify them.
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.axhline(130,color='red')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
# If we also want to identify batsmen whose average is more than 30, we can draw a vertical line at 30.
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.axhline(130,color='red')
plt.axvline(30,color='red')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
# We can draw as many lines as we want.
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.axhline(130,color='red')
plt.axvline(30,color='red')
plt.axhline(140,color='green')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```
