```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
batters=pd.read_csv('batter.csv')
batters.shape
```

So, there are 605 batsman and to plot graph we will 

```python
sample_df=batters.head(100).sample(25,random_state=5)
```

We will take the top 100 batsmen and then randomly select 25 batsmen from them.


```python
sample_df
```

```python
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
```

Now, each dot on the graph represents a batsman, but we cannot tell which dot represents which batsman. This problem can be solved using annotations.

Using annotations, we can place text on the graph to identify specific points.


```python
# Eg:-
x=[1,2,3,4]
y=[5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'Point1')
plt.text(2,6,'Point2')
plt.text(3,7,'Point3',fontdict={'size':12,'color':'brown'})
plt.text(4,8,'Point4')
```

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'])
# Since there are 25 batsmen, we don't want to write plt.text() 25 times. Instead, we will use a loop to add annotations to all the batsmen.
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

We can also change the size of the dots. For example, the batsman who scored more runs can have a larger dot, while the batsman who scored fewer runs can have a smaller dot.

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
# Since there are 25 batsmen, we don't want to write plt.text() 25 times. Instead, we will use a loop to add annotations to all the batsmen.
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])    
```
