Heatmap

A heatmap is basically a way of representing a grid as a graph, where different colors show how the values are distributed across the grid.
It is generally used to compare two categorical variables and visualize their values side by side.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Question: I want to create a grid where the y-axis represents the over number (1–20) and the x-axis represents the ball number (1–6). The grid should show the number of sixes hit on each ball of each over across all IPL matches.


```python
delivery=pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
delivery['ballnumber'].unique()
```

But we only want values from 1 to 6, so...

```python
temp_df=delivery[(delivery['ballnumber'].isin([1,2,3,4,5,6]))&(delivery['batsman_run']==6)]
```

```python
grid=temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')
```

```python
# Now, we have to plot a heatmap.
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.colorbar()
```

Here, the over starts from 0, but we can also remove it like this:

```python
plt.figure(figsize=(20,10))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(),list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
```
