Plot() function:

plot() is a function available in Pandas that allows us to create Matplotlib graphs directly using Pandas.
It makes plotting much easier and reduces a lot of the work involved in creating graphs.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# On a series
s=pd.Series([1,2,3,4,5,6,7])
s.plot(kind='line')
```

```python
# It can also be used on a DataFrame.
import seaborn as sns 
tips=sns.load_dataset('tips')
tips.head()
```

```python
tips['size']=tips['size']*100
tips.head()
```

```python
tips.plot(kind='scatter',x='total_bill',y='tip',title='Cost Analysis',color='red',marker='+',figsize=(10,6),s='size')
```

```python
# If we also want to differentiate the data based on gender, then
tips.plot(kind='scatter',x='total_bill',y='tip',title='Cost Analysis',marker='+',figsize=(10,6),s='size',c='sex',colormap='viridis')
```

```python
# Bar Chart
temp=pd.read_csv('batsman_season_record.csv')
temp.head()
```

```python
temp.plot(kind='bar')
```

```python
# Single bar chart
temp['2015'].plot(kind='bar')
```

```python
tips.groupby('sex')['total_bill'].mean().plot(kind='bar')
```

```python
# Stacked Bar Chart
temp.plot(kind='bar',stacked=True)
```

```python
# Pie 
df=pd.DataFrame(
    {
        'batsman':['Dhawan','Rohit','Kohli','SKY','Pandya','Pant'],
        'match1':[120,90,35,45,12,10],
        'match2':[0,1,123,130,34,45],
        'match3':[50,24,145,45,10,90]
    }
)
df.head()
```

```python
# Single Column
df['match1'].plot(kind='pie',labels=df['batsman'].values,autopct='%0.1f%%',explode=[0,0,0,0.1,0,0])
```

```python
len(df['match1'])
```

```python
# Multiple Pie Chart
df[['match1','match2','match3']].plot(kind='pie',subplots=True,figsize=(15,8))
```

```python
tips.groupby('sex')['tip'].sum()
```

```python
tips.groupby('sex')['tip'].sum().plot(kind='pie',autopct='%0.1f%%')
```

```python
# On Multiindex Dataframes
tips.pivot_table(index='day',columns='sex',values='total_bill',aggfunc='mean').plot(kind='line')
```

```python
tips.pivot_table(index=['day','time'],columns='sex',values='total_bill',aggfunc='mean').plot(kind='line')
```

```python
tips.pivot_table(index=['day','time'],columns=['sex','smoker'],values='total_bill',aggfunc='mean').plot(kind='line')
```
