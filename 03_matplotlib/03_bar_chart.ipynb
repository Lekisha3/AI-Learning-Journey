Bar Chart :-

The x-axis contains the categorical column, and the y-axis displays an aggregated value (e.g., average, sum, or count).

1. Bivariate Analysis or Univariate
2. Numerical (generally an aggregate function) Vs Categorical
3. Use Case: Aggregate analysis of groups 

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns  #to use for dataset
```

```python
# Simple bar chart 
children =[10,20,50,30,40]
colors=['red','green','blue','pink','yellow']

plt.bar(colors,children)
```

```python
# color,label and title
plt.bar(colors,children,color='black')
plt.xlabel('colors')
plt.ylabel('children')
plt.show()
```

```python
#Horizontal bar chart
plt.barh(colors,children,color='black')
'''If the number of categories is more, we can use a horizontal bar chart.
 If the number of categories is less, we can use a vertical bar chart.'''
```

```python
# Multiple bar charts 
df= pd.read_csv('batsman_season_record.csv')
df
```

```python
# so , now if we want to see the bar chart of batsman run in 2015
plt.bar(df['batsman'],df['2015'])
```

```python
# Now, if we want the 2016 graph to appear side by side with the 2015 graph
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
# Doing this will result in overlapping bars, providing no benefit.
```

```python
# To avoid overlapping, we will adjust the width of the bars.
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='blue')
```

```python
# If you want to display the batsman name on the x-axis instead of numbers.
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color='blue')
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# Problem 
'''One problem with a bar chart is that...'''
children=[10,20,40]
colors=['red red red red red red red','blue blue blue blue blue blue blue','black black black black']
plt.bar(colors,children)

# The names on the x-axis are too long, so they overlap and reduce the readability of the chart.
```

```python
# So, to avoid this 
children=[10,20,40]
colors=['red red red red red red red','blue blue blue blue blue blue blue','black black black black']
plt.bar(colors,children)
plt.xticks(rotation='vertical')
plt.show()

# Now, the names will be displayed vertically.
```

```python
# Stacked bar chart 
'''Now, if we want to see the total runs scored by each batsman 
and also want to know how many runs were scored in 2015, 2016, and 2017 using different colors.'''
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2017']+df['2016']),label='2015')
plt.legend()
plt.show()
```
