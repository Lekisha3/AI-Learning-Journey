```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns  #to use for dataset
```

```python
# Simple data
data=[32,45,56,10,15,27,61]
plt.hist(data)
plt.show()
```

A histogram is created by automatically determining the bins, but we can also specify the bins manually.

```python
plt.hist(data,bins=[10,25,40,55,70])
```

We can also reduce the bin size to see the data more clearly.

```python
plt.hist(data,bins=[10,20,25,30,40,50,55,70])
```

```python
# Logarithmic Scale 
arr=np.load('big-array.npy')
plt.hist(arr,bins=[10,20,30,40,50,60,70])
plt.show()
```

Now, here comes the problem: this DataFrame has almost 12,000 data points, and if one range contains that much data, the other ranges may look like they have almost no data.

So, the solution is to use a logarithmic scale instead of a linear scale.

```python
arr=np.load('big-array.npy')
plt.hist(arr,bins=[10,20,30,40,50,60,70],log=True)
plt.show()
```
