```python
#From a pandas dataframe 
batsman = pd.read_csv("sharma-kohli.csv")
batsman   

plt.plot(batsman['index'],batsman["V Kohli"])
```

```python
# plotting multiple plots 
plt.plot(batsman['index'],batsman["V Kohli"])
plt.plot(batsman['index'],batsman["RG Sharma"])
```

```python
# labels title 
plt.plot(batsman['index'],batsman["V Kohli"])
plt.plot(batsman['index'],batsman["RG Sharma"])

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
# Color , Line and marker 
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red')
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dotted")
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dotted",linewidth=5)
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
#The points plotted on a graph are called markers. When these markers are connected, they form the graph. We can also customize or manipulate these markers by changing their size, shape
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dashed",linewidth=5,marker ="D")
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot", marker =">")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
#legend 
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dashed",linewidth=5,marker ="D",label='Virat Kohli')
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot", marker =">",label ="Rohit Sharma")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")
```

```python
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dashed",linewidth=5,marker ="D",label='Virat Kohli')
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot", marker =">",label ="Rohit Sharma")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")

plt.legend()
```

```python
plt.plot(batsman['index'],batsman["V Kohli"],color = 'red',linestyle ="dashed",linewidth=5,marker ="D",label='Virat Kohli')
plt.plot(batsman['index'],batsman["RG Sharma"], color ="#F527CF",linestyle = "dashdot", marker =">",label ="Rohit Sharma")

plt.title("Rohit Sharma Vs Virat Kohli career comparison")
plt.xlabel("season")
plt.ylabel("runs scored")

plt.legend(loc="lower right")
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000, 3500000]
year =[2015,2016,2017,2018,2019,2020,2021]
plt.plot(year,price)
```

```python
#We can perform trimming in this case.
price = [48000,54000,57000,49000,47000,45000, 3500000]
year =[2015,2016,2017,2018,2019,2020,2021]
plt.ylim(0,100000)
plt.plot(year,price)
```

```python
price = [48000,54000,57000,49000,47000,45000, 3500000]
year =[2015,2016,2017,2018,2019,2020,2021]
plt.ylim(0,100000)
plt.xlim(2015,2019)
plt.plot(year,price)
```

```python
#Grid (improve graph readability)
price = [48000,54000,57000,49000,47000,45000, 3500000]
year =[2015,2016,2017,2018,2019,2020,2021]
plt.ylim(0,100000)
plt.xlim(2015,2019)
plt.plot(year,price)
plt.grid()
```
